## 📺 Step 4: Display Generated Video

Find your output `.mp4` file in the directory and run this cell to play it directly in Colab!

```python
import glob
import os
from IPython.display import HTML
from base64 import b64encode

# Get the latest generated mp4 file
videos = glob.glob("*.mp4")
if videos:
    latest_video = max(videos, key=os.path.getctime)
    print(f"Displaying: {latest_video}")
    
    mp4 = open(latest_video, 'rb').read()
    data_url = "data:video/mp4;base64," + b64encode(mp4).decode()
    display(HTML(f"""
    <video width=640 controls>
          <source src="{data_url}" type="video/mp4">
    </video>
    """))
else:
    print("No generated video found yet!")
```
